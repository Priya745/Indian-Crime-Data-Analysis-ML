# Crime Data Analysis using Indian District-wise Crime Data

## 1. Data Understanding

In this first phase, we will load our dataset, understand its structure, and inspect its basic properties. We are using standard Data Science libraries for this.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style for our charts later
sns.set_theme(style="whitegrid")
print("Libraries imported successfully!")
```

### Load the Dataset
We load the CSV file using Pandas. The dataset contains absolute reported crime counts (not population-adjusted rates).

```python
# Load the dataset
file_path = '../data/factors/crime/01_District_wise_crimes_committed_IPC_2001_2012.csv'
df = pd.read_csv(file_path)

# Display shape
print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns\n")

# Display column names
print("Column Names:")
print(list(df.columns))
```

```python
# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Display last 5 rows (useful to spot summary rows at the end of the file)
print("\nLast 5 rows:")
display(df.tail())
```

```python
# Display data types of each column
print("Data Types:")
display(df.dtypes)
```

```python
# Display descriptive statistics for numerical columns
print("\nDescriptive Statistics:")
display(df.describe())
```

### Identifying Data Types and Unique Values
Let's see how many states and years are covered, and categorize our columns into text (categorical) and numbers (numerical).

```python
# Check number of unique states/UTs
unique_states = df['STATE/UT'].nunique()
print(f"Number of unique States/UTs: {unique_states}")

# Check years available
years = sorted(df['YEAR'].unique())
print(f"Years available: {years[0]} to {years[-1]} (Total {len(years)} years)")

# Identify numerical and categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

print(f"\nCategorical Columns ({len(categorical_cols)}): {categorical_cols}")
print(f"Numerical Columns ({len(numerical_cols)}): {numerical_cols}")

# Identify all crime-category columns (all numerical columns except YEAR and TOTAL IPC CRIMES)
crime_cols = [col for col in numerical_cols if col not in ['YEAR', 'TOTAL IPC CRIMES']]
print(f"\nCrime-category Columns ({len(crime_cols)}):")
print(crime_cols)
```

## 2. Data Quality Check

Before analyzing data, we must ensure it is clean. We will check for missing values, duplicates, and mathematically impossible negative values. We will also investigate if there are "summary rows" mixed in with our district-level data.

```python
# Check missing values
missing_values = df.isnull().sum().sum()
print(f"Total missing values in the dataset: {missing_values}")

# Check duplicate rows
duplicate_rows = df.duplicated().sum()
print(f"Total exact duplicate rows: {duplicate_rows}")

# Check for negative values in crime columns
negative_values = (df[crime_cols] < 0).sum().sum()
print(f"Total negative values in crime counts: {negative_values}")
```

### Inspecting Summary Rows
Sometimes datasets contain state-wide totals embedded directly inside the `DISTRICT` column. If we aggregate the data (e.g., sum all districts to find the state total) while these summary rows are present, we will count every crime **twice**. Let's check for rows labeled `TOTAL` or `DELHI UT TOTAL`.

```python
# Find districts that contain the word 'TOTAL'
summary_districts = df[df['DISTRICT'].str.contains('TOTAL', case=False, na=False)]['DISTRICT'].unique()
print(f"Summary rows found in DISTRICT column: {list(summary_districts)}")

# Count how many of these summary rows exist
total_summary_rows = df['DISTRICT'].isin(['TOTAL', 'DELHI UT TOTAL']).sum()
print(f"Number of summary rows to be removed: {total_summary_rows}")
```

### Verifying Mathematical Relationships (Subcategories)
Some columns are subsets of larger columns. Let's verify if the math adds up perfectly.
1. `CUSTODIAL RAPE` + `OTHER RAPE` = `RAPE`
2. `AUTO THEFT` + `OTHER THEFT` = `THEFT`
3. `KIDNAPPING AND ABDUCTION OF WOMEN AND GIRLS` + `KIDNAPPING AND ABDUCTION OF OTHERS` = `KIDNAPPING & ABDUCTION`

We will not delete these columns yet, but we are documenting them so we don't accidentally count them twice during EDA.

```python
# Test 1: Rape
rape_diff = (df['RAPE'] - (df['CUSTODIAL RAPE'] + df['OTHER RAPE'])).abs().sum()
print(f"Difference in Rape formula: {rape_diff} (If 0, it matches perfectly)")

# Test 2: Theft
theft_diff = (df['THEFT'] - (df['AUTO THEFT'] + df['OTHER THEFT'])).abs().sum()
print(f"Difference in Theft formula: {theft_diff}")

# Test 3: Kidnapping
kidnap_diff = (df['KIDNAPPING & ABDUCTION'] - 
               (df['KIDNAPPING AND ABDUCTION OF WOMEN AND GIRLS'] + df['KIDNAPPING AND ABDUCTION OF OTHERS'])).abs().sum()
print(f"Difference in Kidnapping formula: {kidnap_diff}")
```

## 3. Data Preprocessing

Now we will clean the dataset. 
* **Step 1:** We will create a new DataFrame (`df_clean`) that excludes the summary rows. This ensures we do not modify the original data.
* **Step 2:** We will verify the rows are gone.
* **Step 3:** We will re-check for any basic quality issues.
* **Step 4:** We will explicitly document our subcategory and main category columns.

```python
# Step 1: Remove summary rows
print(f"Rows before filtering: {df.shape[0]}")

# Keep rows where DISTRICT is NOT 'TOTAL' and NOT 'DELHI UT TOTAL'
df_clean = df[~df['DISTRICT'].isin(['TOTAL', 'DELHI UT TOTAL'])].copy()

print(f"Rows after filtering: {df_clean.shape[0]}")
```

```python
# Step 2: Verify summary rows are removed
remaining_summary = df_clean[df_clean['DISTRICT'].isin(['TOTAL', 'DELHI UT TOTAL'])]
print(f"Summary rows remaining in clean dataset: {remaining_summary.shape[0]}")
```

```python
# Step 3: Recheck missing, duplicates, and negatives on clean dataset
print(f"Missing values: {df_clean.isnull().sum().sum()}")
print(f"Duplicate rows: {df_clean.duplicated().sum()}")
print(f"Negative values: {(df_clean[crime_cols] < 0).sum().sum()}")
```

```python
# Step 4: Separate and Document Columns
# We will just create lists to keep our columns organized for EDA.

categorical_cols = ['STATE/UT', 'DISTRICT']
numerical_cols = [col for col in df_clean.columns if col not in categorical_cols]

# The total column
summary_total_col = 'TOTAL IPC CRIMES'

# The subcategories we found earlier
subcategory_cols = [
    'CUSTODIAL RAPE', 
    'OTHER RAPE',
    'AUTO THEFT', 
    'OTHER THEFT',
    'KIDNAPPING AND ABDUCTION OF WOMEN AND GIRLS', 
    'KIDNAPPING AND ABDUCTION OF OTHERS'
]

# The main crime categories (excluding subcategories, YEAR, and TOTAL)
main_crime_cols = [c for c in crime_cols if c not in subcategory_cols]

print("Column Groupings Documented:")
print(f"- Main Crime Categories: {len(main_crime_cols)} columns")
print(f"- Subcategories: {len(subcategory_cols)} columns")
print(f"- Summary/Total: '{summary_total_col}'")
```

## 4. Exploratory Data Analysis (EDA)
Now that our data is clean and strictly district-level, we can begin analyzing crime patterns across India.

### 4.1 Year-wise Analysis
Let's look at how the total reported crime volume has changed across India from 2001 to 2012.

```python
# Group by year and sum the total IPC crimes
yearly_crimes = df_clean.groupby('YEAR')['TOTAL IPC CRIMES'].sum().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=yearly_crimes, x='YEAR', y='TOTAL IPC CRIMES', marker='o', linewidth=2.5)
plt.title('Total Reported Crime Volume in India (2001 - 2012)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Reported IPC Crimes', fontsize=12)
plt.xticks(yearly_crimes['YEAR'])
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

**What this shows:** We can observe the overall trend in the volume of reported IPC crimes over the decade.

### 4.2 State-wise Analysis
Let's see which States/UTs reported the highest and lowest total crime volumes across the entire 12-year period.
*(Note: These are absolute counts. Larger states naturally report higher volumes because of larger populations).*

```python
# Group by State and sum the total IPC crimes
state_crimes = df_clean.groupby('STATE/UT')['TOTAL IPC CRIMES'].sum().sort_values(ascending=False).reset_index()

# Plot Top 10 States
plt.figure(figsize=(12, 6))
sns.barplot(data=state_crimes.head(10), x='TOTAL IPC CRIMES', y='STATE/UT', palette='Reds_r', hue='STATE/UT')
plt.title('Top 10 States by Total Reported Crime Volume (2001-2012)', fontsize=14)
plt.xlabel('Total Reported IPC Crimes', fontsize=12)
plt.ylabel('State / UT', fontsize=12)
plt.show()

# Plot Bottom 5 States
plt.figure(figsize=(10, 4))
sns.barplot(data=state_crimes.tail(5), x='TOTAL IPC CRIMES', y='STATE/UT', palette='Blues_r', hue='STATE/UT')
plt.title('Lowest 5 States/UTs by Total Reported Crime Volume (2001-2012)', fontsize=14)
plt.xlabel('Total Reported IPC Crimes', fontsize=12)
plt.ylabel('State / UT', fontsize=12)
plt.show()
```

### 4.3 District-wise Analysis
Let's identify the specific districts across the country that reported the highest volume of crimes overall.

```python
# Group by District and State
district_crimes = df_clean.groupby(['STATE/UT', 'DISTRICT'])['TOTAL IPC CRIMES'].sum().sort_values(ascending=False).reset_index()

# Plot Top 15 Districts
plt.figure(figsize=(12, 6))
sns.barplot(data=district_crimes.head(15), x='TOTAL IPC CRIMES', y='DISTRICT', palette='magma',hue='DISTRICT')
plt.title('Top 15 Districts by Total Reported Crime Volume (2001-2012)', fontsize=14)
plt.xlabel('Total Reported IPC Crimes', fontsize=12)
plt.ylabel('District', fontsize=12)
plt.show()
```

### 4.4 Crime-category Analysis
Which types of crimes are reported most frequently? We will use the `main_crime_cols` list to avoid double-counting the subcategories.

```python
# Sum all main crime categories across the entire dataset
crime_totals = df_clean[main_crime_cols].sum().sort_values(ascending=False).reset_index()
crime_totals.columns = ['Crime Category', 'Total Reported Cases']

# Since 'OTHER IPC CRIMES' is a massive catch-all bucket, let's plot it, but also pay attention to specific crimes.
plt.figure(figsize=(12, 8))
sns.barplot(data=crime_totals.head(10), x='Total Reported Cases', y='Crime Category', palette='viridis',hue='Crime Category')
plt.title('Top 10 Most Frequently Reported Crime Categories (2001-2012)', fontsize=14)
plt.xlabel('Total Reported Cases', fontsize=12)
plt.ylabel('Crime Category', fontsize=12)
plt.show()
```

### 4.5 Individual Crime Trends
Let's see how specific major crimes (`MURDER`, `RAPE`, `THEFT`, `KIDNAPPING & ABDUCTION`) evolved over time. We will use a separate chart for `THEFT` since its volume is much higher than the others, which would skew the Y-axis.

```python
# Group specific crimes by year
crimes_over_time = df_clean.groupby('YEAR')[['MURDER', 'RAPE', 'KIDNAPPING & ABDUCTION']].sum().reset_index()

# Plot Murder, Rape, and Kidnapping
plt.figure(figsize=(10, 5))
for col in ['MURDER', 'RAPE', 'KIDNAPPING & ABDUCTION']:
    sns.lineplot(data=crimes_over_time, x='YEAR', y=col, marker='o', label=col)
    
plt.title('Trends of Violent Crimes (2001-2012)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Reported Cases', fontsize=12)
plt.xticks(crimes_over_time['YEAR'])
plt.legend()
plt.show()

# Plot Theft separately due to scale
theft_over_time = df_clean.groupby('YEAR')['THEFT'].sum().reset_index()
plt.figure(figsize=(10, 4))
sns.lineplot(data=theft_over_time, x='YEAR', y='THEFT', marker='s', color='orange')
plt.title('Trend of Reported Theft Cases (2001-2012)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Reported Cases', fontsize=12)
plt.xticks(theft_over_time['YEAR'])
plt.show()
```

### 4.6 State vs Crime Analysis (Heatmap)
Let's compare how different states experience different types of crimes. We will look at a cross-section of Murder, Rape, Robbery, and Theft for the Top 15 states (by total crime).

```python
# Get top 15 states by total crime
top_states_list = state_crimes.head(15)['STATE/UT'].tolist()

# Filter data for top states and select specific crimes
state_crime_matrix = df_clean[df_clean['STATE/UT'].isin(top_states_list)]
state_crime_matrix = state_crime_matrix.groupby('STATE/UT')[['MURDER', 'RAPE', 'ROBBERY', 'THEFT']].sum()

# We normalize the data (convert to percentages within each state) so we can compare the *distribution*
# rather than absolute numbers, making the heatmap much more readable.
state_crime_percent = state_crime_matrix.div(state_crime_matrix.sum(axis=1), axis=0) * 100

plt.figure(figsize=(10, 8))
sns.heatmap(state_crime_percent, annot=True, fmt=".1f", cmap="YlGnBu", linewidths=.5)
plt.title('Proportion of Selected Crimes within Top 15 States (%)', fontsize=14)
plt.ylabel('State / UT', fontsize=12)
plt.xlabel('Crime Category', fontsize=12)
plt.show()
```

### 4.7 Outlier Analysis
An outlier is a data point that differs significantly from other observations. Here, we will use a Box Plot to look at the distribution of `MURDER` cases across districts. Each dot outside the "box" represents a district that had an unusually high number of reported murders compared to the national average.

```python
plt.figure(figsize=(12, 4))
sns.boxplot(x=df_clean['MURDER'], color='red')
plt.title('Distribution of Murder Cases Across Districts (Outlier Analysis)', fontsize=14)
plt.xlabel('Reported Murder Cases per District-Year', fontsize=12)
plt.show()
```

### 4.8 Correlation Analysis
Correlation measures how strongly two variables are related. 
* A positive correlation (near 1.0) means as Crime A goes up, Crime B also goes up.
* A negative correlation (near -1.0) means as Crime A goes up, Crime B goes down.

We will exclude subcategories and `TOTAL IPC CRIMES` to prevent mathematical relationships from inflating our correlations artificially.

```python
# Select a few major crimes for correlation
corr_crimes = ['MURDER', 'ATTEMPT TO MURDER', 'RAPE', 'KIDNAPPING & ABDUCTION', 
               'ROBBERY', 'BURGLARY', 'THEFT', 'RIOTS', 'CHEATING']

# Calculate correlation matrix
corr_matrix = df_clean[corr_crimes].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, square=True)
plt.title('Correlation Matrix of Major Crime Categories', fontsize=14)
plt.show()
```

### 4.9 Additional Useful Analysis: Violent vs. Property Crime Proportion
**Question:** Over the entire decade, what proportion of these selected major crimes were "Violent" versus "Property/Financial"?
**Why it's useful:** It gives a high-level summary of the nature of reported crime in India.
**Visualization:** A Pie Chart.

```python
violent_crimes_list = ['MURDER', 'ATTEMPT TO MURDER', 'CULPABLE HOMICIDE NOT AMOUNTING TO MURDER', 
                       'RAPE', 'KIDNAPPING & ABDUCTION', 'RIOTS', 'HURT/GREVIOUS HURT', 'DOWRY DEATHS']
property_crimes_list = ['DACOITY', 'ROBBERY', 'BURGLARY', 'THEFT', 'CRIMINAL BREACH OF TRUST', 'CHEATING']

total_violent = df_clean[violent_crimes_list].sum().sum()
total_property = df_clean[property_crimes_list].sum().sum()

plt.figure(figsize=(6, 6))
plt.pie([total_violent, total_property], labels=['Violent Crimes', 'Property & Financial Crimes'], 
        autopct='%1.1f%%', colors=['#ff9999','#66b3ff'], startangle=90, shadow=True)
plt.title('Proportion of Violent vs. Property Crimes (2001-2012)', fontsize=14)
plt.show()
```

## 5. EDA Findings

Based on our objective Exploratory Data Analysis, here are the key findings from the 2001–2012 dataset:

**Important Limitation:** All findings are based on *absolute reported crime counts*. Because we do not have population data, larger states naturally report higher absolute numbers. These findings reflect total volume, not per-capita crime rates.

1. **Year-wise Trends:** The total reported volume of IPC crimes in India showed a steady upward trend from 2001 to 2012. While crimes like Murder remained relatively flat, Kidnapping & Abduction saw a visible increase over the decade.
2. **State-wise Analysis:** Madhya Pradesh, Maharashtra, and Tamil Nadu recorded the highest total reported volume of IPC crimes in this dataset. Conversely, island territories like Lakshadweep and D & N Haveli recorded the lowest volumes.
3. **District-wise Analysis:** Certain districts reported vastly higher total crime volumes than others, indicating that crime reporting is heavily concentrated in specific geographical hotspots (often urban centers).
4. **Crime Category Analysis:** `THEFT` is by far the most frequently reported specific crime category, overshadowing other crimes so significantly that it required an independent scale for time-series visualization. The catch-all category `OTHER IPC CRIMES` makes up the largest segment of the data.
5. **State vs Crime Focus:** The heatmap revealed that the *proportion* of specific crimes varies by state. For example, within the top 15 states, Theft consistently makes up the vast majority of the selected major crimes, but the ratio of Robbery to Theft fluctuates regionally.
6. **Outliers:** The district-level box plot for Murder shows several significant outliers—specific districts that reported drastically higher absolute numbers of murders in a single year compared to the rest of the dataset.
7. **Correlations:** There are strong positive correlations between property crimes (e.g., Robbery and Burglary at `0.72`, Burglary and Theft at `0.77`). Violent crimes like Murder and Attempt to Murder also show a strong positive correlation (`0.79`). This suggests that districts experiencing high volumes of one type of property crime likely experience high volumes of others as well.
